# Nilai Gamma Aktual Model SVM

Notebook ini membaca nilai γ aktual dari model final M1, M2, dan M3 yang tersimpan pada folder `Documents/Skripsi/SVM_bertahap`. Model disetel dengan γ = "scale", sehingga nilai aktualnya dihitung oleh scikit-learn dari jumlah fitur dan variansi fitur latih saat pelatihan. Hasilnya digunakan untuk melengkapi Subbab 5.3.3.

## 1. Lokasi Berkas Model

```python
from pathlib import Path
import joblib
import pandas as pd

ARTIFACT_DIR = Path.home() / "Documents" / "Skripsi" / "SVM_bertahap"
MODEL_FILES = {
    "M1": "svm_m1_raw.pkl",
    "M2": "svm_m2_raw_glcm.pkl",
    "M3": "svm_m3_raw_glcm_hog_lbp.pkl",
}

for model, nama in MODEL_FILES.items():
    path = ARTIFACT_DIR / nama
    print(f"{model}  {nama:30s} {'ditemukan' if path.exists() else 'TIDAK DITEMUKAN'}")
```

## 2. Pembacaan Nilai Gamma

```python
rows = []
for model, nama in MODEL_FILES.items():
    clf = joblib.load(ARTIFACT_DIR / nama)
    rows.append({
        "Model": model,
        "Berkas": nama,
        "Dimensi Fitur": clf.n_features_in_,
        "C": clf.C,
        "Gamma Setelan": clf.gamma,
        "Gamma Aktual": clf._gamma,
    })

hasil = pd.DataFrame(rows)
hasil.style.format({"Gamma Aktual": "{:.4e}"}).hide(axis="index")
```

## 3. Format Penulisan untuk Laporan

```python
SUPERSCRIPT = str.maketrans("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹")

def format_ilmiah(nilai):
    mantisa, eksponen = f"{nilai:.2e}".split("e")
    return f"{mantisa.replace('.', ',')} × 10{str(int(eksponen)).translate(SUPERSCRIPT)}"

for _, r in hasil.iterrows():
    print(f"{r['Model']}: γ = {format_ilmiah(r['Gamma Aktual'])}")
```
